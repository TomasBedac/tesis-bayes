```python
#import stan
import pandas as pd
import numpy as np
import scipy.stats as stats 
import matplotlib
import matplotlib.pyplot as plt
```

```python
# They integrate to 1 
b0 = pd.read_csv('data/bottom_interpolated.csv', names=['gn1','density'],header=None )
j0 = pd.read_csv('data/light_interpolated.csv', names=['gn1','density'], header=None )
c0 = pd.read_csv('data/charm_interpolated.csv', names=['gn1','density'], header=None )
# reshape true curves to m bins
vec = np.linspace(0,99,100)
myrows = [int(x) for x in vec]
jtrue = j0.iloc[myrows]
ctrue = c0.iloc[myrows]
btrue = b0.iloc[myrows]
jtrue.reset_index(drop=True, inplace=True)
ctrue.reset_index(drop=True, inplace=True)
btrue.reset_index(drop=True, inplace=True)
jtrue['density'] = jtrue['density']/sum(jtrue['density'])
ctrue['density'] = ctrue['density']/sum(ctrue['density'])
btrue['density'] = btrue['density']/sum(btrue['density'])
myjets = {'j': jtrue, 'c': ctrue, 'b': btrue}
```

```python
plt.plot(jtrue['gn1'],jtrue['density'], label='j')
plt.plot(ctrue['gn1'],ctrue['density'], label='c')
plt.plot(btrue['gn1'],btrue['density'], label='b')
for i in range(0,100,10):
    plt.axvline(x=jtrue['gn1'][i], linestyle='dashed', linewidth=0.5, color='black')
plt.legend()
plt.xlabel('GN1')
plt.show()
```

```python
# bin at which the b-efficiency is 77%
sum(btrue['density'][51:])
```

## ROC curves

#### Sampling and threshold

```python
np.random.choice(len(jtrue['density']), 4,p=jtrue['density'])
```

```python
N = 10**7
j = []
b = []
for i in range(N):
    if i % 100000 == 0: print(i)
    j.append(min(np.random.choice(len(jtrue['density']), 4,p=jtrue['density'])))
    b.append(min(np.random.choice(len(btrue['density']), 4,p=btrue['density'])))
```

```python
hj = plt.hist(j, density=1, histtype='step', color='red', bins=np.linspace(0,100,30), label='min of jjjj')
hb = plt.hist(b, density=1, histtype='step', color='blue', bins=np.linspace(0,100,30), label='min of bbbb')
plt.plot(jtrue['density'], 'r:', label='j')
plt.plot(btrue['density'], 'b:', label='b')
plt.axvline(x=51, linestyle='dashed', color='black', label='77% efficiency')
plt.xlabel('Min of four draws')
plt.legend()
plt.show()
```

The ratio of pp -> hh -> 4b cross-section (0.0046 fb) to pp -> jjjj (6*10E6 fb) is 10E-9.  Hence one should compare the red and green areas on the right of the black-dashed 77% efficiency threshold.  

If for instance one expects a signal count of N_S ~ 10 events, then to have s/sqrt(B) ~ 1, we need N_B ~ 100, or more suitable to say N_B/N_S ~ 10.  Therefore the areas on the right of the 77%-line times their respetive xsection should have this ratio.  Therefore

$$ A_{red} \cdot 10^6 \approx 10 \cdot A_{blue} \cdot 10^{-3} $$

Therefore

$$ \frac{A_{red}}{A_{blue}} \approx 10^{-8}$$

Which could be the case.  One should compute much better the red curve...

```python
# Actually:
len([x for x in j if x>51])/len(j)
# hence we may be close to the value 10^-8 if >51 
# It looks that this run may have had a fluctuation, the maximum values is 59 with 10^7 events
```

```python
roc = []
for i in range(len(hj[0])):
    roc.append([sum(hj[0][i:])*(hj[1][2]-hj[1][1]),sum(hb[0][i:])*(hb[1][2]-hb[1][1])])
```

```python
plt.plot([x[0] for x in roc], [x[1] for x in roc],'-bo')
plt.title('ROC curve')
roc_auc = -np.trapz( [x[1] for x in roc], [x[0] for x in roc])
print('roc_auc= ',roc_auc)
plt.grid(alpha=0.4)
plt.show()
```
