```python
#import stan
import pandas as pd
import numpy as np
import scipy.stats as stats 
import matplotlib
import matplotlib.pyplot as plt
```

### Individual components

```python
# They integrate to 1 
b0 = pd.read_csv('data/bottom_interpolated.csv', names=['gn1','density'],header=None )
j0 = pd.read_csv('data/light_interpolated.csv', names=['gn1','density'], header=None )
c0 = pd.read_csv('data/charm_interpolated.csv', names=['gn1','density'], header=None )
# reshape true curves to m bins
vec = np.linspace(0,99,100)
myrows = [int(x) for x in vec]
jtrue = j0.iloc[myrows]
ctrue = c0.iloc[myrows]
btrue = b0.iloc[myrows]
jtrue.reset_index(drop=True, inplace=True)
ctrue.reset_index(drop=True, inplace=True)
btrue.reset_index(drop=True, inplace=True)
jtrue['density'] = jtrue['density']/sum(jtrue['density'])
ctrue['density'] = ctrue['density']/sum(ctrue['density'])
btrue['density'] = btrue['density']/sum(btrue['density'])
myjets = {'j': jtrue, 'c': ctrue, 'b': btrue}
```

```python
plt.plot(jtrue['gn1'],jtrue['density'], label='j')
plt.plot(ctrue['gn1'],ctrue['density'], label='c')
plt.plot(btrue['gn1'],btrue['density'], label='b')
for i in range(0,100,10):
    plt.axvline(x=jtrue['gn1'][i], linestyle='dashed', linewidth=0.5, color='black')
plt.legend()
plt.xlabel('GN1')
plt.show()
```

```python
myjets['j']
```

### The classes

```python
# Xsections in pb at 14 TeV
cut = 1
# QCD
sjjjj = 6459000 * cut
scccc = 549 * cut
sbbbb = 566 * cut
sbbcc = 1118 * cut
sccjj = 177400 * cut
sbbjj = 182800 * cut
# EW
stth = 0.246
szh = 0.030
sbbh = 0.016
# Signal
#shh = 0.011 # https://arxiv.org/pdf/1610.07922, table 46, page 189
#shhbbbb = shh * 0.569**2
shhbbbb = 0.0046 # Leandro calculation (12DEC2024 email)
```

```python
def eff(jet, threshold):
    return sum(myjets[jet]['density'][threshold:])
```

```python
eff('b',60) #65%
```

```python
eff('b',51) #77% of b
```

```python
eff('b',40) #87%
```

```python
eff('b',25) #97% of b
```

```python
events = []
lum = 3000000 # 3000 fb = 3*10E6 pb
#lum = 150000
for mythreshold in range(1,99):
    events.append([mythreshold,
                sjjjj * lum * eff('j',mythreshold)**4,                
                scccc * lum * eff('c',mythreshold)**4,
                sbbbb * lum * eff('b',mythreshold)**4,
                sccjj * lum * eff('c',mythreshold)**2 * eff('j',mythreshold)**2,
                sbbjj * lum * eff('b',mythreshold)**2 * eff('j',mythreshold)**2,
                sbbcc * lum * eff('b',mythreshold)**2 * eff('c',mythreshold)**2,
                stth * lum * eff('b',mythreshold)**4,
                szh * lum * eff('b',mythreshold)**4,
                sbbh * lum * eff('b',mythreshold)**4,
                shhbbbb * lum * eff('b',mythreshold)**4])

events = pd.DataFrame(events, columns = ['threshold','jjjj','cccc','bbbb','ccjj','bbjj','bbcc','tth', 'Zh', 'hbb', 'signal'])
```

```python
events[-20:]
```

##### checks

```python
plt.plot(events['bbbb']/events['signal'], label='bbbb/signal')
plt.plot(events['bbcc']/events['signal'], label='bbcc/signal')
plt.plot(events['cccc']/events['signal'], label='cccc/signal')
#plt.plot(events['ccjj']/events['signal'], label='jjcc/signal')
plt.legend()
plt.show()
```

### Fraction of events

```python
events_frac = []
for j in range(len(events)):
    tot = sum(events.iloc[j][1:])
    events_frac.append([events.iloc[j][0], 
                        events.iloc[j][1]/tot,
                         events.iloc[j][2]/tot,
                         events.iloc[j][3]/tot,
                         events.iloc[j][4]/tot,
                         events.iloc[j][5]/tot,
                         events.iloc[j][6]/tot,
                         events.iloc[j][7]/tot,
                         events.iloc[j][8]/tot,
                         events.iloc[j][9]/tot,
                         events.iloc[j][10]/tot])

events_frac = pd.DataFrame(events_frac, columns = ['threshold','jjjj','cccc','bbbb','ccjj','bbjj','bbcc','tth', 'Zh', 'hbb', 'signal'])
```

```python
events_frac
```

```python
for col in list(events_frac)[1:]:
    plt.plot(events_frac['threshold'], events_frac[col], label=col)
plt.legend()
plt.show()
```

```python
x = events_frac.iloc[:, 0]
y = events_frac.iloc[:, 1:]
befficiencies = [round(eff('b',int(x[i])),2) for i in range(1,98)] + [0]

# Plot
fig, ax = plt.subplots(figsize=(14, 10))

ax.stackplot(x, y.T, labels=y.columns)


# Axvlines with corresponding text
lines_positions = [60, 40, 51, 25]
eff_values = [0.65, 0.87, 0.77, 0.97]

for pos, effs in zip(lines_positions, eff_values):
    ax.axvline(x=pos, color='black', linestyle='dashed')
    ax.text(pos - 0.5, 0.5, f'$\epsilon(b)={effs}$', rotation=90, va='center', ha='right', fontsize=13)


# Label the legend, axes, and title
ax.set_xticks(x, befficiencies, rotation=60, fontsize=7)
ax.legend(loc='upper left')
ax.set_xlabel('b-efficiency')
ax.set_ylabel('Fractions')
ax.set_title('Class abundance as function of pre-selection b-tagging threshold')
#plt.savefig('preselection.png')
plt.show()
```

### $S/\sqrt(B)$

```python
events
```

```python
events.iloc[0][1:10]
```

```python
SosqrtB = []
SoB = []
for row in range(len(events)):
    SosqrtB.append(events.iloc[row]['signal'] / np.sqrt(sum(events.iloc[row][1:10])))
    SoB.append(events.iloc[row]['signal'] / sum(events.iloc[row][1:10]))
events['S/sqrt(B)'] = SosqrtB
events['S/B'] = SoB
```

```python
plt.plot(events['signal'], events['S/sqrt(B)'], label='$S/\sqrt{B}$')
#plt.plot(events['signal'], events['S/B'], label='$S/B$')
plt.xlabel('Signal events')
plt.ylabel('$S/\sqrt{B}$')
plt.title('Lum = '+str(lum/1000)+' fb-1')
plt.legend()
plt.show()
```

```python
events_frac['b_eff'] = np.zeros(len(events_frac))

for i in range(len(events_frac)):
    events_frac.loc[i,'b_eff'] = eff('b',i)
```
